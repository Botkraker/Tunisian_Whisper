```python
import torch, librosa
from transformers import WhisperForConditionalGeneration, WhisperProcessor

name = "TuniSpeech-AI/whisper-tunisian-dialect"
processor = WhisperProcessor.from_pretrained(name)
model = WhisperForConditionalGeneration.from_pretrained(name)

audio, sr = librosa.load("sound.wav", sr=16000)
inputs = processor(audio, sampling_rate=16000, return_tensors="pt")

with torch.no_grad():
    ids = model.generate(inputs.input_features, language="arabic", task="transcribe")

print(processor.batch_decode(ids, skip_special_tokens=True)[0])
```
